# CheckAtlas examples : Evaluate and compare different atlases

In this example, we show how to run checkatlas in a folder and produce a summary html file. The three atlases are downloaded in Scanpy (.h5ad) format.

WARNING: Install and run only MultiQC from https://github.com/becavin-lab/MultiQC/tree/checkatlas. Otherwise checkatlas files will not be taken into account.

## Download datasets

Datasets are downloaded from the cellxgene server. This might takes few minutes.

B-cells compartment<br>
From: <a href=https://cellxgene.cziscience.com/collections/62ef75e4-cbea-454e-a0ce-998ec40223d3>Cross-tissue immune cell analysis reveals tissue-specific features in humans - Domínguez Conde et al. (2022) Science</a>

```python
%%bash
cd /home/bernadettem/bernadettenotebook/Ritwik/checkatlas_project/atlas_data

# Download B-cells compartment (482 MB)
# Dataset ID: 3bf9cc6c-9acc-4f63-a03d-34e2e9953aea
curl -L -o B-cells_compartment.h5ad \
  "https://datasets.cellxgene.cziscience.com/3bf9cc6c-9acc-4f63-a03d-34e2e9953aea.h5ad"
```

Endothelial<br>
From: <a href=https://cellxgene.cziscience.com/collections/e5f58829-1a66-40b5-a624-9046778e74f5>Tabula Sapiens - The Tabula Sapiens Consortium* et al. (2022) Science</a>

```python
%%bash
cd /home/bernadettem/bernadettenotebook/Ritwik/checkatlas_project/atlas_data

# Download Tabula Sapiens Endothelial
# Dataset ID: 89cb5969-86ee-4cca-a1c9-f4c1bab19085
curl -L -o Tabula_Sapiens_Endothelial.h5ad \
  "https://datasets.cellxgene.cziscience.com/89cb5969-86ee-4cca-a1c9-f4c1bab19085.h5ad"
```

Fetal<br>
From: 
<a href=https://cellxgene.cziscience.com/collections/13d1c580-4b17-4b2e-85c4-75b36917413f>Single cell derived mRNA signals across human kidney tumors - Young et al. (2021) Nat Commun</a>

```python
%%bash
cd /home/bernadettem/bernadettenotebook/Ritwik/checkatlas_project/atlas_data

# Download Fetal kidney
# Dataset ID: 5d80925e-efbe-46cd-b3a7-2033d0d85e12
curl -L -o Fetal.h5ad \
  "https://datasets.cellxgene.cziscience.com/5d80925e-efbe-46cd-b3a7-2033d0d85e12.h5ad"
```

## Run checkatlas

If checkatlas is installed in your environment, you just need to run this cell. 

This will produce all metric tables and figures needed.

```python
# Set environment variable to use conda's C++ library
import os

os.environ["LD_LIBRARY_PATH"] = (
    f"/home/bernadettem/anaconda3/envs/ritwik_base/lib:{os.environ.get('LD_LIBRARY_PATH', '')}"
)

# Verify the library path
print(f"LD_LIBRARY_PATH set to: {os.environ['LD_LIBRARY_PATH']}")
```
