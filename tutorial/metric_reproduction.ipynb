## Smart Column Detection for Post-Hoc Quality Assessment

Based on CheckAtlas framework and single-cell best practices, we need to identify:

1. **Reference/Ground Truth Annotations** - Typically from:
   - `cell_type` (standardized CELLxGENE schema)
   - `author_cell_type` (original curator annotations)
   - Other expert-curated labels

2. **Predicted/Computed Annotations** - From clustering algorithms:
   - `seurat_clusters` (Seurat clustering)
   - `leiden` (Leiden algorithm)
   - `louvain` (Louvain algorithm)
   - `annotCell` (automated annotation tools)

The metrics compare **reference annotations** vs **predicted clusters/annotations** to assess quality.


`But`, A hardcoded knowledge base won't scale for real-world deployment. Let me design a robust, intelligent column detection strategy that works globally using multiple heuristics and ML-based approaches.

### Unified Detection Strategy: CheckAtlasColumnDetector

**NEW APPROACH**: Single detector for all CheckAtlas metric types

#### Key Features:

1. **Regex-Based Pattern Matching** - All patterns use regular expressions for flexible, global matching
2. **Unified Predicted Annotations** - Same detection logic for both:
   - Annotation metrics (predicted vs reference comparison)
   - Clustering metrics (cluster quality assessment)
3. **Comprehensive Scanning** - Single detector scans:
   - `.obs` columns for annotations (reference + predicted/clusters)
   - `.obsm` keys for embeddings (PCA, UMAP, t-SNE)
4. **Domain-Aware Scoring**:
   - Reference annotations: 40% semantic + 60% statistical
   - Predicted annotations: 60% semantic + 40% statistical
   - Embeddings: Semantic patterns + shape validation

#### Advantages Over Hardcoded Knowledge Base:

✅ **Globally Scalable** - Works on any h5ad file following single-cell standards  
✅ **Production-Ready** - No manual column mapping needed  
✅ **Task-Agnostic** - Detects parameters for multiple metric types from one scan  
✅ **Confidence Scoring** - Provides transparency on detection quality  
✅ **Standards-Compliant** - Follows CheckAtlas, CELLxGENE, and HCA conventions


```python
class CheckAtlasColumnDetector:
    """
    Unified column detector for all CheckAtlas metric types.

    Detects parameters for:
    1. Annotation metrics (reference vs predicted annotations)
    2. Clustering metrics (embeddings + cluster labels)
    3. Other metric types

    Uses multiple strategies:
    - Semantic analysis (regex pattern matching on column names)
    - Statistical properties (cardinality, distribution)
    - Data quality indicators (completeness, consistency)
    - Domain-specific heuristics (single-cell biology standards)
    """

    def __init__(self, adata):
        self.adata = adata
        self.obs_columns = adata.obs.columns.tolist()
        self.obsm_keys = list(adata.obsm.keys()) if hasattr(adata, "obsm") else []
        self.analysis_results = {}

    def analyze_column_semantics(self, col_name: str) -> Dict[str, float]:
        """
        Score column name based on semantic indicators using regex patterns.
        Returns confidence scores for different column types.
        """
        col_lower = col_name.lower()

        # Define semantic patterns with confidence weights
        semantic_scores = {
            "reference_annotation": 0.0,
            "predicted_annotation": 0.0,  # For both clusters AND automated predictions
            "metadata": 0.0,
        }

        # Reference annotation indicators (ground truth)
        reference_patterns = {
            r"\bcell[_\s]?type\b": 1.0,  # cell_type, celltype
            r"\bauthor[_\s]?cell[_\s]?type\b": 0.95,  # author_cell_type
            r"\bann(otation)?[_\s]?(finest|level|label)\b": 0.9,  # ann_finest_level
            r"\bground[_\s]?truth\b": 0.95,
            r"\breference\b": 0.85,
            r"\blabel\b": 0.6,
            r"\bontology\b": 0.7,
            r"\bcurated\b": 0.8,
        }

        predicted_patterns = {
            # Cluster-based (from Seurat, Scanpy, etc.)
            r"\b(leiden|louvain)\b": 1.0,
            r"\bseurat[_\s]?clusters?\b": 1.0,
            r"\bcluster(s|ing)?\b": 0.9,
            r"\bRNA[_\s]?snn[_\s]?res": 0.95,
            r"\bSCT[_\s]?snn[_\s]?res": 0.95,
            r"\bgraph[_\s]?clust": 0.9,
            r"\bkmeans\b": 0.85,
            # Automated prediction-based
            r"\bpredict(ed)?\b": 1.0,
            r"(?i).*celltypist.*": 1.0,
            r"\binfer(red)?\b": 0.95,
            r"\bannot[_\s]?cell\b": 0.9,  # annotCell
            r"\bauto(matic|mated)?\b": 0.85,
            r"\bsctype\b": 0.9,
            r"\bcelltypist\b": 0.9,
            r"\bsingleR\b": 0.9,
        }

        # Metadata indicators (to exclude)
        metadata_patterns = {
            r"\b(batch|sample|donor|patient|replicate)\b": 1.0,
            r"(?i).*(id|dataset).*": 1.0,
            r"\b(tissue|organ|disease|condition)\b": 0.7,
            r"\b(barcode|cell[_\s]?id|obs[_\s]?names)\b": 1.0,
            r"\b(n[_\s]?(genes|counts|umis))\b": 0.9,
            r"\b(percent|pct)[_\s]": 0.8,
            r"\b(phase|cycle)\b": 0.7,
        }

        # Calculate scores
        for pattern, weight in reference_patterns.items():
            if re.search(pattern, col_lower):
                semantic_scores["reference_annotation"] = max(
                    semantic_scores["reference_annotation"], weight
                )

        for pattern, weight in predicted_patterns.items():
            if re.search(pattern, col_lower):
                semantic_scores["predicted_annotation"] = max(
                    semantic_scores["predicted_annotation"], weight
                )

        for pattern, weight in metadata_patterns.items():
            if re.search(pattern, col_lower):
                semantic_scores["metadata"] = max(semantic_scores["metadata"], weight)

        return semantic_scores

    def analyze_obsm_semantics(self) -> List[str]:
        """
        Simple regex-based matching for all obsm embedding keys.
        Returns list of all matched embedding keys.
        """
        # Embedding patterns (simple list of regex patterns)
        embedding_patterns = [
            r"^(X_pca|X_umap|X_tsne|X_scvi)$",
            r"^X_diffmap$",
            r"^X_draw_graph",
            r"^X_.*pca",
            r"^X_.*umap",
            r"^X_.*tsne",
            r"^X_",  # Match any X_ prefixed key
        ]

        matched_keys = []

        for key in self.obsm_keys:
            for pattern in embedding_patterns:
                if re.search(pattern, key):
                    matched_keys.append(key)
                    break  # Move to next key once matched

        return matched_keys

    def analyze_column_statistics(self, col_name: str) -> Dict[str, float]:
        """
        Analyze statistical properties of the column data.
        """
        data = self.adata.obs[col_name]

        stats_profile = {
            "cardinality_ratio": 0.0,  # unique values / total values
            "completeness": 0.0,  # non-null ratio
            "is_categorical": 0.0,  # likely categorical data
            "label_length_avg": 0.0,  # avg string length
            "numeric_ratio": 0.0,  # % numeric values
            "n_unique": 0,  # absolute number of unique values
        }

        # Completeness
        non_null = data.notna().sum()
        total = len(data)
        stats_profile["completeness"] = non_null / total if total > 0 else 0

        # Cardinality
        n_unique = data.nunique()
        stats_profile["n_unique"] = n_unique
        stats_profile["cardinality_ratio"] = n_unique / total if total > 0 else 0

        # Check if categorical
        if data.dtype == "object" or data.dtype.name == "category":
            stats_profile["is_categorical"] = 1.0

            # Average label length (for string data)
            valid_data = data.dropna()
            if len(valid_data) > 0:
                try:
                    avg_len = valid_data.astype(str).str.len().mean()
                    stats_profile["label_length_avg"] = avg_len
                except:
                    pass
        else:
            # Check if numeric data could be cluster IDs
            try:
                numeric_data = pd.to_numeric(data, errors="coerce")
                stats_profile["numeric_ratio"] = numeric_data.notna().sum() / total
            except:
                pass

        return stats_profile

    def score_reference_annotation(self, col_name: str) -> float:
        """
        Calculate confidence score for reference/ground truth annotation.

        Scoring: Semantic 40% + Statistical 60%
        """
        semantic = self.analyze_column_semantics(col_name)
        stats = self.analyze_column_statistics(col_name)

        score = 0.0

        # Semantic contribution (40%)
        score += semantic["reference_annotation"] * 0.4

        # Penalize if it looks like metadata
        score -= semantic["metadata"] * 0.6

        # Statistical indicators (60%)
        # 1. Must be categorical
        if stats["is_categorical"] < 0.5:
            return 0.0

        # 2. Cardinality: ideal range 5-200 unique types
        cardinality = stats["n_unique"]
        if 5 <= cardinality <= 200:
            score += 0.25
        elif cardinality < 5:
            score += 0.05  # Too few categories
        elif cardinality > 500:
            score -= 0.2  # Too many (likely cell IDs)

        # 3. Completeness
        score += stats["completeness"] * 0.2

        # 4. Label length (descriptive names)
        if stats["label_length_avg"] > 5:
            score += 0.15
        elif stats["label_length_avg"] > 10:
            score += 0.25  # Very descriptive

        return np.clip(score, 0, 1)

    def score_predicted_annotation(self, col_name: str) -> float:
        """
        Calculate confidence score for predicted/cluster annotation.

        Scoring: Semantic 60% + Statistical 40%
        (Higher semantic weight for better pattern matching)
        """
        semantic = self.analyze_column_semantics(col_name)
        stats = self.analyze_column_statistics(col_name)

        score = 0.0

        # Semantic contribution (60%)
        score += semantic["predicted_annotation"] * 0.6

        # Penalize if it looks like metadata
        score -= semantic["metadata"] * 0.6

        # Statistical indicators (40%)

        # 1. Unique elements (cardinality): ideal range 2-100
        cardinality = stats["n_unique"]
        if 2 <= cardinality <= 50:
            score += 0.20  # Optimal range for clusters
        elif 51 <= cardinality <= 100:
            score += 0.15  # Still reasonable for predictions
        elif cardinality < 2:
            return 0.0  # Not useful (single value)
        elif cardinality > 200:
            score -= 0.15  # Too many (likely cell IDs or noise)

        # 2. Completeness (10%)
        score += stats["completeness"] * 0.10

        # 3. Data type compatibility (10%)
        if stats["numeric_ratio"] > 0.9:
            score += 0.10  # Numeric cluster IDs
        elif stats["is_categorical"] > 0.5:
            score += 0.10  # Categorical labels

        return np.clip(score, 0, 1)

    def score_embedding(self) -> List[str]:
        """
        Simple regex-based detection for all embeddings.
        Returns list of all matched embedding keys.
        """
        return self.analyze_obsm_semantics()

    def detect_all_parameters(self) -> Dict:
        """
        UNIFIED DETECTION METHOD

        Detects parameters for ALL CheckAtlas metric types from a single scan:
        - Annotation metrics: reference_annotation + predicted_annotation (.obs)
        - Clustering metrics: embeddings (.obsm) + cluster_labels (.obs)

        Returns comprehensive detection results.
        """
        results = {
            "annotation": {"reference": [], "predicted": []},
            "clustering": {"embeddings": [], "cluster_labels": []},
        }

        # Scan .obs columns
        for col in self.obs_columns:
            # Skip obviously irrelevant columns
            if col.startswith("_") or col in ["index"]:
                continue

            ref_score = self.score_reference_annotation(col)
            pred_score = self.score_predicted_annotation(col)

            if ref_score > 0.3:
                results["annotation"]["reference"].append((col, ref_score))

            if pred_score > 0.3:
                results["annotation"]["predicted"].append((col, pred_score))
                # Same for clustering (cluster labels)
                results["clustering"]["cluster_labels"].append((col, pred_score))

            # Store detailed analysis
            self.analysis_results[col] = {
                "reference_score": ref_score,
                "predicted_score": pred_score,
                "semantics": self.analyze_column_semantics(col),
                "statistics": self.analyze_column_statistics(col),
            }

        # Scan .obsm keys (for clustering embeddings)
        matched_embeddings = self.score_embedding()

        for key in matched_embeddings:
            # Get basic shape info
            embedding_data = self.adata.obsm[key]
            shape = embedding_data.shape
            n_components = embedding_data.shape[1] if len(embedding_data.shape) > 1 else 1

            results["clustering"]["embeddings"].append(
                (key, {"shape": shape, "n_components": n_components})
            )

        # Sort results
        results["annotation"]["reference"].sort(key=lambda x: x[1], reverse=True)
        results["annotation"]["predicted"].sort(key=lambda x: x[1], reverse=True)
        results["clustering"]["cluster_labels"].sort(key=lambda x: x[1], reverse=True)
        # No sorting for embeddings - just return in order found

        return results

    def get_best_parameters(
        self, min_reference_score: float = 0.5, min_predicted_score: float = 0.5
    ) -> Dict:
        """
        Get best parameters for all metric types with unified detection.

        Note: Embeddings are selected by regex match only (first match used).

        Returns:
            Dictionary with recommendations for annotation and clustering metrics
        """
        all_results = self.detect_all_parameters()

        # Annotation metrics parameters
        best_reference = None
        if (
            all_results["annotation"]["reference"]
            and all_results["annotation"]["reference"][0][1] >= min_reference_score
        ):
            best_reference = all_results["annotation"]["reference"][0][0]

        best_predicted = [
            col
            for col, score in all_results["annotation"]["predicted"]
            if score >= min_predicted_score
        ]

        # Clustering metrics parameters - return ALL matched embeddings
        best_embeddings = []
        if all_results["clustering"]["embeddings"]:
            best_embeddings = [
                {"key": emb[0], "metadata": emb[1]}
                for emb in all_results["clustering"]["embeddings"]
            ]

        best_clusters = [
            col
            for col, score in all_results["clustering"]["cluster_labels"]
            if score >= min_predicted_score
        ]

        return {
            "annotation_metrics": {
                "reference": {
                    "column": best_reference,
                    "score": (
                        all_results["annotation"]["reference"][0][1]
                        if all_results["annotation"]["reference"]
                        else 0.0
                    ),
                    "all_candidates": all_results["annotation"]["reference"][:5],
                },
                "predicted": {
                    "columns": best_predicted[:5],
                    "all_candidates": all_results["annotation"]["predicted"][:10],
                },
            },
            "clustering_metrics": {
                "embeddings": best_embeddings,  # Now returns ALL matched embeddings
                "cluster_labels": {
                    "columns": best_clusters[:5],
                    "all_candidates": all_results["clustering"]["cluster_labels"][:10],
                },
            },
        }

    def print_comprehensive_report(self):
        """Print comprehensive detection report for all metric types."""
        results = self.detect_all_parameters()

        print("=" * 80)
        print("CHECKATLAS UNIFIED COLUMN DETECTION REPORT")
        print("=" * 80)

        print(f"\n📋 Total .obs columns analyzed: {len(self.obs_columns)}")
        print(f"📋 Total .obsm keys analyzed: {len(self.obsm_keys)}")

        # Annotation Metrics Section
        print(f"\n{'='*80}")
        print("ANNOTATION METRICS PARAMETERS")
        print(f"{'='*80}")

        print(f"\n✅ REFERENCE ANNOTATION CANDIDATES (Ground Truth):")
        print(f"{'Column Name':<40} {'Score':>8} {'Unique':>8} {'Complete':>10}")
        print("-" * 80)
        for col, score in results["annotation"]["reference"][:10]:
            stats = self.analysis_results[col]["statistics"]
            n_unique = stats["n_unique"]
            completeness = f"{stats['completeness']*100:.1f}%"
            print(f"{col:<40} {score:>8.3f} {n_unique:>8} {completeness:>10}")

        print(f"\n🎯 PREDICTED ANNOTATION CANDIDATES (Clusters/Predictions):")
        print(f"{'Column Name':<40} {'Score':>8} {'Unique':>8} {'Complete':>10}")
        print("-" * 80)
        for col, score in results["annotation"]["predicted"][:10]:
            stats = self.analysis_results[col]["statistics"]
            n_unique = stats["n_unique"]
            completeness = f"{stats['completeness']*100:.1f}%"
            print(f"{col:<40} {score:>8.3f} {n_unique:>8} {completeness:>10}")

        # Clustering Metrics Section
        print(f"\n{'='*80}")
        print("CLUSTERING METRICS PARAMETERS")
        print(f"{'='*80}")

        print(f"\n🗺️  EMBEDDING/REPRESENTATION SPACES (.obsm):")
        print(f"{'Key':<20} {'Shape':<15} {'Components':>12}")
        print("-" * 80)
        if results["clustering"]["embeddings"]:
            for key, meta in results["clustering"]["embeddings"]:
                print(f"{key:<20} {str(meta['shape']):<15} {meta['n_components']:>12}")
        else:
            print("  ⚠️  No embeddings found in .obsm")

        print(f"\n🎯 CLUSTER LABEL COLUMNS (.obs):")
        print(f"{'Column Name':<40} {'Score':>8} {'Unique':>8} {'Complete':>10}")
        print("-" * 80)
        if results["clustering"]["cluster_labels"]:
            for col, score in results["clustering"]["cluster_labels"][:10]:
                stats = self.analysis_results[col]["statistics"]
                n_unique = stats["n_unique"]
                completeness = f"{stats['completeness']*100:.1f}%"
                print(f"{col:<40} {score:>8.3f} {n_unique:>8} {completeness:>10}")
        else:
            print("  ⚠️  No cluster labels found in .obs")

        # Recommendations
        best = self.get_best_parameters()
        print(f"\n{'='*80}")
        print("💡 RECOMMENDATIONS")
        print(f"{'='*80}")

        print(f"\n📊 For ANNOTATION METRICS:")
        print(
            f"   Reference: {best['annotation_metrics']['reference']['column']} "
            + f"(score: {best['annotation_metrics']['reference']['score']:.3f})"
        )
        print(
            f"   Predicted: {', '.join(best['annotation_metrics']['predicted']['columns'][:3]) if best['annotation_metrics']['predicted']['columns'] else 'None found'}"
        )

        print(f"\n📊 For CLUSTERING METRICS:")
        if best["clustering_metrics"]["embeddings"]:
            print(
                f"   Embeddings ({len(best['clustering_metrics']['embeddings'])} found):"
            )
            for emb in best["clustering_metrics"]["embeddings"]:
                meta = emb["metadata"]
                print(f"      • {emb['key']} ({meta['n_components']} components)")
        else:
            print(f"   Embeddings: ⚠️  None found")
        print(
            f"   Cluster Labels: {', '.join(best['clustering_metrics']['cluster_labels']['columns'][:3]) if best['clustering_metrics']['cluster_labels']['columns'] else 'None found'}"
        )
```

### Running Clustering Metrics

Now we'll create a function to run all clustering metrics with intelligently detected parameters.


```python
def run_clustering_metrics_on_atlas(
    adata,
    atlas_name,
    embedding_key,
    cluster_col,
    metrics_to_run=None,
    use_raw_counts=False,
):
    """
    Run all clustering metrics on a single atlas file.

    Args:
        adata: AnnData object
        atlas_name: Name of the atlas
        embedding_key: Key for embedding in adata.obsm (e.g., 'X_pca', 'X_umap')
        cluster_col: Column name for cluster annotations in adata.obs
        metrics_to_run: List of metric names (default: all from cluster.__all__)
        use_raw_counts: If True and no embedding_key, use adata.X

    Returns:
        Dictionary of metric results
    """
    if metrics_to_run is None:
        metrics_to_run = cluster.__all__

    # Get cluster annotations
    annotations = adata.obs[cluster_col]

    # Remove NaN values
    valid_mask = annotations.notna()
    annotations_clean = annotations[valid_mask]

    # Get count representation (embedding or raw counts)
    if embedding_key and embedding_key in adata.obsm.keys():
        count_repr = adata.obsm[embedding_key][valid_mask]
        repr_type = embedding_key
    elif use_raw_counts:
        count_repr = (
            adata.X.toarray()[valid_mask]
            if hasattr(adata.X, "toarray")
            else adata.X[valid_mask]
        )
        repr_type = "Raw counts (adata.X)"
    else:
        return {}  # Silently return empty dict if embedding not found

    # Run metrics silently - main loop handles all printing
    results = {
        "n_clusters": annotations_clean.nunique(),
        "n_valid_samples": len(annotations_clean),
        "n_total_samples": len(annotations),
        "embedding_shape": count_repr.shape,
        "metrics": {},
    }

    for metric_name in metrics_to_run:
        try:
            metric_module = getattr(cluster, metric_name)
            score = metric_module.run(count_repr, annotations_clean)
            results["metrics"][metric_name] = score
        except Exception as e:
            results["metrics"][metric_name] = None

    return results
```

### 🔧 Output Buffering Fix Applied

**Problem:** Python and Jupyter notebook output buffering was causing:
- Print statements appearing out of order
- Duplicate messages from different code sections
- Mixed output from multiple iterations

**Solution Applied:**
1. ✅ **`flush=True` on all print statements** - Forces immediate output
2. ✅ **`sys.stdout.flush()` at strategic points** - Additional flush after:
   - Data loading (scanpy operations)
   - Parameter detection
   - Each metric combination
   - Atlas completion
3. ✅ **`sc.settings.verbosity = 0`** - Disables scanpy's own print messages to prevent duplicate "Loaded" messages

**Result:** Clean, sequential, non-duplicated output! 🎉

```python
# Run clustering metrics on all atlases using UNIFIED detector
clustering_results = {}

# Deduplicate data_files to avoid processing duplicates
unique_data_files = list(set(data_files))

# Disable scanpy verbosity to prevent duplicate messages
sc.settings.verbosity = 0  # 0 = errors only, 1 = warnings, 2 = info, 3 = hints

print("=" * 80, flush=True)
print("RUNNING CLUSTERING METRICS ON ALL ATLASES", flush=True)
print("=" * 80, flush=True)
sys.stdout.flush()  # Force immediate output

for data_file in sorted(unique_data_files):
    print(f"\n{'='*80}", flush=True)
    print(f"📂 PROCESSING: {data_file}", flush=True)
    print(f"{'='*80}", flush=True)
    sys.stdout.flush()  # Force flush before loading data

    try:
        # Load data (scanpy may have its own print statements)
        adata = sc.read_h5ad(os.path.join(data_path, data_file))
        atlas_name = data_file.replace(".h5ad", "")

        # Small delay to ensure file loading completes and buffers clear
        time.sleep(0.1)

        print(
            f"✓ Loaded: {adata.shape[0]:,} cells × {adata.shape[1]:,} genes", flush=True
        )
        sys.stdout.flush()  # Force flush after loading

        # UNIFIED column detection
        detector = CheckAtlasColumnDetector(adata)
        params = detector.get_best_parameters()

        # Extract clustering metrics parameters
        embeddings = params["clustering_metrics"]["embeddings"]
        cluster_cols = params["clustering_metrics"]["cluster_labels"]["columns"]

        # Print detected parameters
        # print(f"\n🔍 DETECTED PARAMETERS:", flush=True)
        # print(f"{'-'*80}", flush=True)
        # print(f"  🗺️  Embeddings: {len(embeddings)} found", flush=True)
        for emb in embeddings:
            meta = emb["metadata"]
            print(
                f"      • {emb['key']}: {meta['n_components']} components, shape={meta['shape']}",
                flush=True,
            )

        print(f"\n  🎯 Cluster Labels: {len(cluster_cols)} found", flush=True)
        for col in cluster_cols[:5]:
            print(f"      • {col}", flush=True)
        if len(cluster_cols) > 5:
            print(f"      ... and {len(cluster_cols) - 5} more", flush=True)
        sys.stdout.flush()  # Force flush after detection

        if not embeddings:
            print(f"\n⚠️  No embeddings found - skipping atlas", flush=True)
            sys.stdout.flush()
            continue

        if not cluster_cols:
            print(f"\n⚠️  No cluster columns found - skipping atlas", flush=True)
            sys.stdout.flush()
            continue

        # Run metrics for ALL embeddings × ALL cluster columns
        atlas_clustering_results = {}

        print(f"\n🚀 RUNNING METRICS:", flush=True)
        print(f"{'-'*80}", flush=True)
        sys.stdout.flush()  # Force flush before metric computation

        total_combinations = len(embeddings) * min(len(cluster_cols), 3)
        current_combo = 0

        for emb in embeddings:
            embedding_key = emb["key"]

            for cluster_col in cluster_cols[:3]:  # Top 3 cluster columns
                current_combo += 1

                print(
                    f"\n  [{current_combo}/{total_combinations}] {embedding_key} × {cluster_col}",
                    flush=True,
                )
                sys.stdout.flush()  # Force flush before metric computation

                metric_results = run_clustering_metrics_on_atlas(
                    adata, atlas_name, embedding_key, cluster_col, use_raw_counts=False
                )

                if not metric_results:
                    print(f"      ⚠️  Skipped - embedding not found", flush=True)
                    sys.stdout.flush()
                    continue

                # Print summary for this combination
                print(
                    f"      Shape: {metric_results['embedding_shape']}, "
                    + f"Clusters: {metric_results['n_clusters']}, "
                    + f"Valid: {metric_results['n_valid_samples']:,}/{metric_results['n_total_samples']:,}",
                    flush=True,
                )

                # Print metric results
                for metric_name, metric_value in metric_results["metrics"].items():
                    if metric_value is not None:
                        print(
                            f"         ✓ {metric_name:<25s} = {metric_value:>8.4f}",
                            flush=True,
                        )
                    else:
                        print(f"         ✗ {metric_name:<25s} = Error", flush=True)

                sys.stdout.flush()  # Force flush after each combination

                # Store with unique key
                if embedding_key not in atlas_clustering_results:
                    atlas_clustering_results[embedding_key] = {}

                atlas_clustering_results[embedding_key][cluster_col] = {
                    "metrics": metric_results["metrics"],
                    "embedding_used": embedding_key,
                    "n_components": emb["metadata"]["n_components"],
                    "n_clusters": metric_results["n_clusters"],
                    "n_valid_samples": metric_results["n_valid_samples"],
                }

        clustering_results[atlas_name] = {
            "embeddings": [emb["key"] for emb in embeddings],
            "results": atlas_clustering_results,
        }

        print(
            f"\n✅ Completed {atlas_name}: {len(embeddings)} embedding(s) × {min(len(cluster_cols), 3)} cluster column(s)",
            flush=True,
        )
        sys.stdout.flush()  # Force flush after atlas completion

    except Exception as e:
        print(f"\n❌ ERROR processing {data_file}: {str(e)}", flush=True)
        sys.stdout.flush()
        import traceback

        traceback.print_exc()
        continue

print(f"\n\n{'='*80}", flush=True)
print(f"✅ CLUSTERING METRIC COMPUTATION COMPLETE", flush=True)
print(f"{'='*80}", flush=True)
print(f"\n📊 Summary:", flush=True)
print(f"   Total atlases processed: {len(clustering_results)}", flush=True)
for atlas_name, data in clustering_results.items():
    n_embeddings = len(data["embeddings"])
    n_combinations = sum(len(clusters) for clusters in data["results"].values())
    print(
        f"   • {atlas_name}: {n_embeddings} embedding(s), {n_combinations} combination(s)",
        flush=True,
    )
print(flush=True)
sys.stdout.flush()  # Final flush
```

```python
def run_annotation_metrics_on_atlas(
    adata, atlas_name, ref_col, pred_col, metrics_to_run=None
):
    """
    Run all annotation metrics on a single atlas file.

    Args:
        adata: AnnData object
        atlas_name: Name of the atlas
        ref_col: Reference annotation column (ground truth)
        pred_col: Predicted annotation column (clusters/automated predictions)
        metrics_to_run: List of metric names (default: all from annot.__all__)

    Returns:
        Dictionary of metric results
    """
    if metrics_to_run is None:
        metrics_to_run = annot.__all__

    # Get the annotations
    reference = adata.obs[ref_col]
    predicted = adata.obs[pred_col]

    # Remove NaN values
    valid_mask = ~(reference.isna() | predicted.isna())
    reference_clean = reference[valid_mask]
    predicted_clean = predicted[valid_mask]

    print(f"\n{'='*80}")
    print(f"Atlas: {atlas_name}")
    print(f"Reference: {ref_col} ({reference_clean.nunique()} unique)")
    print(f"Predicted: {pred_col} ({predicted_clean.nunique()} unique)")
    print(f"Valid samples: {len(reference_clean)} / {len(reference)}")
    print(f"{'='*80}\n")

    results = {}

    for metric_name in metrics_to_run:
        # Handle the fowlkes_mallow typo

        try:
            metric_module = getattr(annot, metric_name)

            # Run the metric
            score = metric_module.run(
                annotation=predicted_clean, ref_annotation=reference_clean
            )

            results[metric_name] = score
            print(f"✓ {metric_name:30s} = {score:.4f}")

        except Exception as e:
            results[metric_name] = None
            print(f"✗ {metric_name:30s} - Error: {str(e)[:50]}")

    return results
```

```python
# Store all results
all_results = {}

print("=" * 80)
print("RUNNING ANNOTATION METRICS ON ALL ATLASES")
print("=" * 80)

for data_file in data_files:
    print(f"\n\n{'#'*80}")
    print(f"# Processing: {data_file}")
    print(f"{'#'*80}")

    # Load data
    adata = sc.read_h5ad(os.path.join(data_path, data_file))
    atlas_name = data_file.replace(".h5ad", "")

    # UNIFIED column detection
    detector = CheckAtlasColumnDetector(adata)
    detection_results = detector.get_best_parameters(
        min_reference_score=1.0, min_predicted_score=0.95
    )

    # Extract annotation metrics parameters
    ref_col = detection_results["annotation_metrics"]["reference"]["column"]
    predicted_cols = detection_results["annotation_metrics"]["predicted"]["columns"]

    if not ref_col:
        print(f"\n⚠️  No suitable reference column found in {data_file}.")
        print(f"   Top candidates were:")
        for col, score in detection_results["annotation_metrics"]["reference"][
            "all_candidates"
        ][:3]:
            print(f"     - {col}: {score:.3f}")
        print(f"   Skipping...")
        continue

    if not predicted_cols:
        print(f"\n⚠️  No suitable predicted columns found in {data_file}.")
        print(f"   Top candidates were:")
        for col, score in detection_results["annotation_metrics"]["predicted"][
            "all_candidates"
        ][:3]:
            print(f"     - {col}: {score:.3f}")
        print(f"   Skipping...")
        continue

    print(f"\n🔍 Detected Columns:")
    print(
        f"   Reference: {ref_col} (score: {detection_results['annotation_metrics']['reference']['score']:.3f})"
    )
    print(f"   Predicted: {predicted_cols}")

    time.sleep(2)

    # Run metrics for each predicted column found
    atlas_results = {}

    for pred_col in predicted_cols[:3]:  # Run on top 3 predicted columns
        comparison_name = f"{atlas_name}:{ref_col}_vs_{pred_col}"

        try:
            metric_results = run_annotation_metrics_on_atlas(
                adata, atlas_name, ref_col, pred_col
            )
            atlas_results[pred_col] = metric_results

        except Exception as e:
            print(f"\n✗ Error processing {comparison_name}: {e}")
            continue

    all_results[atlas_name] = {"reference": ref_col, "metrics": atlas_results}

print(f"\n\n{'='*80}")
print("METRIC COMPUTATION COMPLETE")
print(f"{'='*80}\n")
```

## Results Analysis and Export

```python
def export_results_to_dataframe(all_results: Dict) -> pd.DataFrame:
    """
    Convert nested results dictionary to a clean DataFrame for analysis.
    """
    rows = []

    for atlas_name, atlas_data in all_results.items():
        ref_col = atlas_data["reference"]

        for pred_col, metrics_dict in atlas_data["metrics"].items():
            for metric_name, metric_value in metrics_dict.items():
                rows.append(
                    {
                        "atlas": atlas_name,
                        "reference_column": ref_col,
                        "predicted_column": pred_col,
                        "metric": metric_name,
                        "value": metric_value,
                    }
                )

    return pd.DataFrame(rows)


# Export and analyze results
if all_results:
    results_df = export_results_to_dataframe(all_results)

    print("\n" + "=" * 80)
    print("RESULTS SUMMARY")
    print("=" * 80)

    # Display sample results
    print("\n📊 Sample Results:")
    print(results_df.head(20).to_string())

    # Pivot table for better visualization
    print("\n📈 Metric Scores by Atlas and Prediction Column:")
    pivot = results_df.pivot_table(
        index=["atlas", "predicted_column"],
        columns="metric",
        values="value",
        aggfunc="first",
    )
    print(pivot.to_string())

    # Save to CSV
    output_file = os.path.join(data_path, "checkatlas_metric_reproduction_results.csv")
    results_df.to_csv(output_file, index=False)
    print(f"\n✅ Results saved to: {output_file}")

    # Summary statistics
    print("\n📊 Metric Statistics (across all atlases):")
    metric_stats = results_df.groupby("metric")["value"].agg(
        ["count", "mean", "std", "min", "max"]
    )
    print(metric_stats.to_string())

    # Best performing combinations
    print("\n🏆 Top 5 Best Performing Atlas-Prediction Combinations:")
    avg_scores = (
        results_df.groupby(["atlas", "predicted_column"])["value"]
        .mean()
        .sort_values(ascending=False)
    )
    print(avg_scores.head(5).to_string())

else:
    print("\n⚠️  No results to export. Please run the metric computation first.")
```
